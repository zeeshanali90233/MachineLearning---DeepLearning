# PCA (Principal Component Analysis)

Principal Component Analysis (PCA) is a dimensionality reduction technique that transforms correlated features into a smaller set of uncorrelated variables called principal components while preserving as much variance as possible.

It is commonly used to reduce the number of features, speed up machine learning models, and visualize high-dimensional data.

```python
import pandas as pd
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('Iris.csv')
print("Dataset loaded successfully.")
```

```python
df.head()
```

```python
features = df.iloc[:, :-1]
target = df.iloc[:, -1]


# Apply PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(features)

```

```python
X_pca
```

```python
# Create a DataFrame with the principal components
pca_df = pd.DataFrame(data=X_pca, columns=['Principal Component 1', 'Principal Component 2'])

# Optionally add the target variable for visualization
pca_df['target'] = target.values

print("\nExplained variance ratio per principal component:")
print(pca.explained_variance_ratio_)
print(f"Total explained variance: {sum(pca.explained_variance_ratio_):.2f}")

print("\nDataFrame with Principal Components:")
print(pca_df.head())

```

```python
# Visualize the PCA results (for 2 components)
plt.figure(figsize=(8, 6))
sns.scatterplot(x='Principal Component 1', y='Principal Component 2', hue='target', data=pca_df, palette='viridis', s=100, alpha=0.8)
plt.title('2 Component PCA of IRIS Dataset')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.grid(True)
plt.show()
```
